```python
df = pd.read_csv('/kaggle/input/titanic-datasets/titanic.csv')
```

```python
population = df['Age'].dropna()
sample_age = population.sample(25).values
```

```python
sample_age 
```

Assume that the we know the population mean and it is given as 35, now will setup the two hypothesis 

H0 => Population mean is 35


H1 => population mean is less than 35

```python
#check the normality of the Age Column using Shapiro wilk Test
from scipy.stats import shapiro

shapiro_age = shapiro(sample_age)
print(shapiro_age)
```

## the p_vlaue given by the shapiro test is used to decide whether the data follows normality or Not

### if p_value < 0.05 then data is Not Normal
### if p_value >= 0.05 then data is Normal

```python
#our p_value is 0.9 that is age column follows normal distribution
```

```python
population_mean = 35 #we have assumed 
```

## T - test

```python
import scipy.stats as stats
#applying one sample t test and finding the value of the p_value corresponding to the t test value

#t_value = (sample_mean - population_mean)/(sample_std/sqrt(sample_size))
t_statistics, p_value = stats.ttest_1samp(sample_age, population_mean)  


print("t-statistics: ", t_statistics)
print("p value: ", p_value/2)  

#internally the function is using the 2 tailed test but here we have h1 as
#mean < 35 that is single tail test so we need to divide the p_value with 2 to get the correct value
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")

else:
    print("Fail to Reject the Null Hypothesis")
```
