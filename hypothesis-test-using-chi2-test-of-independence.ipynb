```python
df = pd.read_csv('/kaggle/input/titanic-datasets/titanic.csv')
```

```python
contingency_table = pd.crosstab(df['Survived'], df['Pclass'])
contingency_table
```

```python
#Perform The Chi_2 Test of independence 
from scipy.stats import chi2_contingency

chi2, p_value, df, expected = chi2_contingency(contingency_table)

#Print the results

print("\n Chi_Square Statistics: {: .2f}".format(chi2))
print("\n P Value: {: .4f}".format(p_value))
print("\n Degrees of Freedom: {}".format(df))
print("\n Expected Frequencies: \n{}".format(expected))

```

```python
alpha = 0.05

if p_value < alpha:
    print("We Reject the NULL Hypothesis. There is the relationship between the two columns")
else:
    print("We Fail to Reject the Null Hypothesis. that is there is no significant association between two columns")
```
