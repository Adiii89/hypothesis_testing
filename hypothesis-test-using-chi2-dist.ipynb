```python
df = pd.read_csv('/kaggle/input/titanic-datasets/titanic.csv')
```

```python
df.head()
```

```python
#Count Passengers in each Class
#Observed Frequencies

class_counts = df['Pclass'].value_counts().sort_index()
print(class_counts)
```

```python
#Calculate the Expected Frequency using theoretical Distribustion i.e Uniform Dist.

total_passengers = len(df)
expected_counts = total_passengers/3
expected = [expected_counts]*3  #list 

print("Expeccted Passengers Counts (assuming Uniform Dist): \n", expected)
```

```python
#Perform the Chi Square Goodness of Fit  test
from scipy.stats import chisquare
chi2, p_value = chisquare(class_counts, expected)

#Print the result
print("\nChi Squared Statistics: {:.2f}".format(chi2))
print("P-value: {:.4f}".format(p_value))
```

```python
alpha = 0.05

if p_value < alpha:
    print("\n Reject the Null Hypothesis. The distribution of the passengers among the classes is not uniform.")
else:
    print("\n Fail to Reject the NULL Hypothesis. The Distribution of passengers among the classes is uniform.")
```
